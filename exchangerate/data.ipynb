```python
import pandas as pd

# 1. Read the BAM data directly from Hugging Face
print("Downloading BAM rates...")
url = "https://huggingface.co/datasets/AllRatesToday/central-bank-exchange-rates/resolve/main/rates/bam.csv"
df = pd.read_csv(url, parse_dates=["date"])

# 2. Filter for 2021 to 2025
df = df[(df['date'].dt.year >= 2020) & (df['date'].dt.year <= 2025)]

# 3. Filter only for "reference" rates (ignoring buy/sell/middle)
if 'reference' in df['type'].unique():
    df = df[df['type'] == 'reference']

# 4. Format it for my ingestion script
clean_df = pd.DataFrame({
    'rate_date': df['date'].dt.date,
    'currency_code': df['quote'].str.upper(), # e.g., USD, EUR
    'rate_mad': df['value'],
    'unit': 1
})

# 5. Deduplicate (just in case) and save to bronze
clean_df = clean_df.drop_duplicates(subset=['rate_date', 'currency_code'])


clean_df.to_csv('data/bronze/fx_2020_2025.csv', index=False)
print(f"Success! Saved {len(clean_df)} rows to data/bronze/fx_2020_2025.csv")
```
